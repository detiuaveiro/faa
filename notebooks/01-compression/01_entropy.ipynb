# Lab 01: Information Theory & Compression

In this notebook, we explore empirical Shannon entropy and model complexity via MDL.

```python
import math
from collections import Counter


def shannon_entropy(data: str) -> float:
    if not data:
        return 0.0
    counts = Counter(data)
    total = len(data)
    return -sum((c / total) * math.log2(c / total) for c in counts.values())


print(f"Entropy of repetitive string: {shannon_entropy('ABABABAB'):.4f}")
```
